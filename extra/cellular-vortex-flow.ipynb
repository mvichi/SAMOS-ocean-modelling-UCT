```python
import numpy as np
import matplotlib.pyplot as plt

# Define grid
x = np.linspace(0, 2*np.pi, 40)
y = np.linspace(0, 2*np.pi, 40)
X, Y = np.meshgrid(x, y)

# Define a typical cellular flow (incompressible)
# Streamfunction ψ = sin(x) * sin(y)
# Velocity components: u = dψ/dy, v = -dψ/dx
k = 1  # number of cells per direction
U =  np.sin(k*X) * np.cos(k*Y) # zonal velocity
V = -np.cos(k*X) * np.sin(k*Y) # meridional velocity

# Create figure
plt.figure(figsize=(8, 6))

# Plot streamlines
#plt.streamplot(X, Y, U, V, density=1.5, linewidth=1, arrowsize=1.2)

# Overlay velocity vectors
plt.quiver(X, Y, U, V, color='black', alpha=0.8)

# Labels and title
plt.title("Example of Cellular Vortex Flow")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 2*np.pi)
plt.ylim(0, 2*np.pi)

plt.axis('equal')
plt.tight_layout()
plt.savefig('cellular_flow_field_4.png')
```
